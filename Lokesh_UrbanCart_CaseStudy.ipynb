```python
def get_product_price(product_id, products_list):
    for product in products_list:
        if product['Product_ID'] == product_id:
            return product['Price']

    return None

get_product_price("P08",products)
```

```python
def get_customer_name(customer_id, customers_list):
    for customer in customers_list:
        if customer['Customer_ID'] == customer_id:
            return customer['Name']

get_customer_name("C01",customers)
```

```python
def calculate_order_value(price, quantity):
    return price * quantity
calculate_order_value(4999, 2)
```

```python
def classify_order(value, high_limit=3000, low_limit=1000):
    if value > high_limit:
        return "High"
    elif value > low_limit:
        return "Medium"
    else:
        return "Low"

classify_order(4000)
```

```python
def generate_invoice(order, products_list, customers_list):

    product_price = get_product_price(order['Product_ID'], products_list)
    customer_name = get_customer_name(order['Customer_ID'], customers_list)
    order_value = calculate_order_value(product_price, order['Quantity'])

    invoice = f"""
-------------------------------
          INVOICE
-------------------------------
Order ID      : {order['Order_ID']}
Customer Name : {customer_name}
Product ID    : {order['Product_ID']}
Quantity      : {order['Quantity']}
Price         : ₹{product_price}
Order Value   : ₹{order_value}
Status        : {order['Status']}
-------------------------------
"""

    return invoice

print(generate_invoice(orders[0],products,customers))
```

```python
def calculate_total_revenue(orders_list, products_list):
    total_revenue = 0

    for order in orders_list:

        if order['Status'] == 'Cancelled':
            continue

        price = get_product_price(order['Product_ID'], products_list)
        order_value = calculate_order_value(price, order['Quantity'])

        total_revenue += order_value

    return total_revenue

calculate_total_revenue(orders,products)
```

```python
def top_customer_by_spend(orders_list, products_list, customers_list):
    customer_spending = {}

    for order in orders_list:


        if order['Status'] != 'Delivered':
            continue

        price = get_product_price(order['Product_ID'], products_list)
        order_value = calculate_order_value(price, order['Quantity'])

        customer_id = order['Customer_ID']

        if customer_id not in customer_spending:
            customer_spending[customer_id] = 0

        customer_spending[customer_id] += order_value


    top_customer_id = None
    highest_spend = 0

    for customer_id in customer_spending:
        if customer_spending[customer_id] > highest_spend:
            highest_spend = customer_spending[customer_id]
            top_customer_id = customer_id

    return get_customer_name(top_customer_id, customers_list)

top_customer = top_customer_by_spend(orders, products, customers)

print("Top Customer:", top_customer)
```

```python
def apply_offer(price, *discounts):
    for discount in discounts:
        price = price - (price * discount / 100)

    return price
print(apply_offer(1000, 10, 5))
```

#MODULE 6

```python
# ==========================================
# URBANCART BUSINESS SUMMARY REPORT
# ==========================================

# 1. Total number of products
total_products = len(products)

# Distinct categories
unique_categories = set()

for product in products:
    unique_categories.add(product['Category'])

total_categories = len(unique_categories)


# 2. Total customers + Premium / Regular split
total_customers = len(customers)

premium_count = 0
regular_count = 0

for customer in customers:
    if customer['Membership'] == 'Premium':
        premium_count += 1
    else:
        regular_count += 1


# 3. Total orders + status split
total_orders = len(orders)

delivered_count = 0
pending_count = 0
cancelled_count = 0

for order in orders:
    if order['Status'] == 'Delivered':
        delivered_count += 1
    elif order['Status'] == 'Pending':
        pending_count += 1
    elif order['Status'] == 'Cancelled':
        cancelled_count += 1


# 4. Total Revenue and 5. Average Order Value
total_revenue = calculate_total_revenue(orders, products)

average_order_value = total_revenue / delivered_count


# 6. Top Customer
top_customer = top_customer_by_spend(orders, products, customers)


# 7. Number of out-of-stock products
out_of_stock = 0

for product in products:
    if product['Stock'] == 0:
        out_of_stock += 1


# 8. Most popular category among Delivered orders
category_counts = {}

for order in orders:

    if order['Status'] != 'Delivered':
        continue

    for product in products:

        if order['Product_ID'] == product['Product_ID']:

            category = product['Category']

            if category not in category_counts:
                category_counts[category] = 0

            category_counts[category] += 1


# Find category with highest count
most_popular_category = None
highest_count = 0

for category in category_counts:

    if category_counts[category] > highest_count:
        highest_count = category_counts[category]
        most_popular_category = category


# ==========================================
# FINAL REPORT
# ==========================================

print("\n" + "=" * 50)
print("        URBANCART BUSINESS SUMMARY REPORT")
print("=" * 50)

print("\nPRODUCTS")
print("-" * 50)
print(f"Total Products              : {total_products}")
print(f"Distinct Categories         : {total_categories}")

print("\nCUSTOMERS")
print("-" * 50)
print(f"Total Customers             : {total_customers}")
print(f"Premium Members             : {premium_count}")
print(f"Regular Members             : {regular_count}")

print("\nORDERS")
print("-" * 50)
print(f"Total Orders                : {total_orders}")
print(f"Delivered                   : {delivered_count}")
print(f"Pending                     : {pending_count}")
print(f"Cancelled                   : {cancelled_count}")

print("\nFINANCIAL PERFORMANCE")
print("-" * 50)
print(f"Total Revenue               : ₹{total_revenue:,.2f}")
print(f"Average Order Value         : ₹{average_order_value:,.2f}")
print(f"Top Customer                : {top_customer}")

print("\nINVENTORY & PRODUCT INSIGHTS")
print("-" * 50)
print(f"Out-of-Stock Products       : {out_of_stock}")
print(f"Most Popular Category       : {most_popular_category}")

print("\n" + "=" * 50)
print("             END OF REPORT")
print("=" * 50)
```
